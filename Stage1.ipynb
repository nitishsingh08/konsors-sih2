# CREATING **TIMESTAMP**

```python
combined_df['datetime'] = pd.to_datetime(
    combined_df['acq_date'].astype(str) + ' ' +
    combined_df['acq_time'].astype(str).str.zfill(4),
    format='%Y-%m-%d %H%M',
    errors='coerce'
)
```

```python
combined_df['date'] = combined_df['datetime'].dt.date
```

```python
import geopandas as gpd

gdf = gpd.GeoDataFrame(
    combined_df,
    geometry=gpd.points_from_xy(
        combined_df.longitude,
        combined_df.latitude
    ),
    crs='EPSG:4326'
)
```

```python
print(gdf.shape)
print(gdf.crs)
print(gdf[['latitude', 'longitude', 'datetime']].head())
```

```python
import numpy as np
from pyproj import Transformer

# Transformer: WGS84 lat/lon -> EPSG:6933
transformer = Transformer.from_crs(
    "EPSG:4326",
    "EPSG:6933",
    always_xy=True
)

# Convert directly to projected X/Y
x, y = transformer.transform(
    combined_df['longitude'].to_numpy(),
    combined_df['latitude'].to_numpy()
)

print(x[:5])
print(y[:5])

```

```python
# Remove only completely identical rows #abhi remove krne ki jarurat nhi hai baki ke components jaruri hai

combined_df = combined_df.drop_duplicates().reset_index(drop=True)

print("Remaining records:", len(combined_df))
```

```python
duplicate_count = combined_df.duplicated(
    subset=duplicate_cols,
    keep=False
).sum()

print(duplicate_count)
```

```python
combined_df["x"] = x
combined_df["y"] = y
```

```python
combined_df
```

```python
SPATIAL_THRESHOLD = 1000       # metres
TEMPORAL_THRESHOLD = 12        # hours
```

```python
TEMPORAL_THRESHOLD_SECONDS = (
    TEMPORAL_THRESHOLD * 60 * 60
)
```

```python
combined_df["grid_x"] = np.floor(
    combined_df["x"] / SPATIAL_THRESHOLD
).astype("int32")

combined_df["grid_y"] = np.floor(
    combined_df["y"] / SPATIAL_THRESHOLD
).astype("int32")
```

```python
combined_df["time_seconds"] = (
    combined_df["datetime"].astype("int64") // 10**9
)
```

```python
combined_df["time_bin"] = (
    combined_df["time_seconds"] //
    TEMPORAL_THRESHOLD_SECONDS
).astype("int64")
```
